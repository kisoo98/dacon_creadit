```python
cat_cols = train.select_dtypes(include=['object', 'string']).columns
num_cols = train.select_dtypes(exclude=['object', 'string']).columns
```

```python
train['age'] = -train['DAYS_BIRTH'] // 365
train['employment_years'] = train['DAYS_EMPLOYED'].apply(
    lambda x: -x // 365 if x < 0 else 0
)
```

```python
train = train.drop(columns=['index', 'DAYS_BIRTH', 'DAYS_EMPLOYED'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 깨짐 방지 (Windows)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 1️⃣ 타겟(credit) 분포
plt.figure(figsize=(6,4))
sns.countplot(x='credit', data=train)
plt.title('신용도(credit) 클래스 분포')
plt.show()

# 2️⃣ 나이 분포 (DAYS_BIRTH → age)
train['age'] = -train['DAYS_BIRTH'] // 365

plt.figure(figsize=(6,4))
sns.histplot(train['age'], bins=30, kde=True)
plt.title('나이 분포')
plt.show()

# 3️⃣ 직업(occyp_type) 결측치 확인
plt.figure(figsize=(6,4))
sns.countplot(
    y=train['occyp_type'].fillna('Unknown'),
    order=train['occyp_type'].fillna('Unknown').value_counts().index
)
plt.title('직업 유형 분포 (결측치 포함)')
plt.show()
```
