###데이터로드

```python
import pandas as pd
import numpy as np

train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')

print(train.shape, test.shape)
```

###타겟 분리

```python
TARGET = 'credit'

y = train[TARGET]
train = train.drop(columns=[TARGET])
```

파생 변수 확정

```python
#나이, 근속연수
train['age'] = (-train['DAYS_BIRTH'] // 365).astype(int)
test['age'] = (-test['DAYS_BIRTH'] // 365).astype(int)

def convert_employment_days(x):
    return -x // 365 if x < 0 else 0

train['employment_years'] = train['DAYS_EMPLOYED'].apply(convert_employment_days)
test['employment_years'] = test['DAYS_EMPLOYED'].apply(convert_employment_days)
```

```python
#가족1인당 소득
train['income_per_person'] = train['income_total'] / train['family_size']
test['income_per_person'] = test['income_total'] / test['family_size']
```

```python
DROP_COLS = ['index', 'DAYS_BIRTH', 'DAYS_EMPLOYED']
#안쓰는 칼럼 제거
train = train.drop(columns=DROP_COLS)
test = test.drop(columns=DROP_COLS)
```

```python
#결측치 처리
train['occyp_type'] = train['occyp_type'].fillna('Unknown')
test['occyp_type'] = test['occyp_type'].fillna('Unknown')
```

```python
#범주형/ 수치형분리
cat_cols = train.select_dtypes(include=['object', 'string']).columns
num_cols = train.select_dtypes(exclude=['object', 'string']).columns

print(cat_cols)
print(num_cols)
```

```python
cat_features = list(cat_cols) #CatBoost 평가
```

```python
#복구/저장

train[TARGET] = y

train.to_csv('../data/train_processed.csv', index=False)
test.to_csv('../data/test_processed.csv', index=False)
```
