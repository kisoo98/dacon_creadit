```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python

train = pd.read_csv('../data/train_processed.csv')
```

```python
sns.countplot(x='credit', data=train)
plt.title('Target distribution (credit)')
plt.show()
```

```python
num_cols = ['age', 'income_total', 'income_per_person', 'employment_years']
train[num_cols].hist(bins=30, figsize=(10,8))
plt.tight_layout()
plt.show()
```

```python
for col in ['occyp_type', 'income_type', 'family_type']:
    sns.countplot(y=col, data=train,
                  order=train[col].value_counts().index)
    plt.title(col)
    plt.show()
```

```python
sns.boxplot(x='credit', y='income_per_person', data=train)
plt.title('credit vs income_per_person')
plt.show()
```

- credit 2 클래스의 비율이 가장 높아 클래스 불균형이 존재함

- 소득 관련 변수는 credit에 따라 분포 차이가 관찰됨
